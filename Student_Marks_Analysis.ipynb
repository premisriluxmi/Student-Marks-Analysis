# Student Marks Analysis

## Objective
To analyze student marks using Python and calculate the total, average, highest, and lowest marks.

```python
# Student marks
marks = [78, 85, 92, 67, 74, 88, 95, 81, 69, 90]

marks
```

```python
total = sum(marks)

total
```

```python
average = total / len(marks)

average
```

```python
highest = max(marks)

highest
```

```python
lowest = min(marks)

lowest
```

```python
print("Student Marks Analysis")
print("----------------------")
print("Total Marks   :", total)
print("Average Marks :", average)
print("Highest Mark  :", highest)
print("Lowest Mark   :", lowest)
```

## Conclusion

The student marks were successfully analyzed using Python.

- Total marks: **819**
- Average marks: **81.9**
- Highest mark: **95**
- Lowest mark: **67**

This analysis demonstrates how Python lists and built-in functions such as `sum()`, `max()`, `min()`, and `len()` can be used to perform basic student data analysis.
